```python
import sys
sys.path.insert(0, '../src')
import matplotlib.pyplot as plt
import lab01

raw = lab01.load_data('../data/cafe_sales.csv')
clean = lab01.clean_data(raw)
raw.head()
```

## Plot 1 — Missing values per column (raw data)

```python
# ============ TODO (Plot 1) ============
# Bar chart of lab01.summarize_missing(raw).
# Label both axes and give the figure a title.
# =======================================

missing = lab01.summarize_missing(raw)
missing.plot(kind="bar")

plt.xlabel("Column")
plt.ylabel("Number of missing values")
plt.title("Missing values per column")
plt.tight_layout()
plt.savefig("fig1.png", dpi=150)
plt.show()
```

## Plot 2 — Distribution of `total_price` (clean data)

```python
# ============ TODO (Plot 2) ============
# Histogram of clean['total_price'] (~30 bins).
# Briefly note in a comment: is it skewed? why?
# =======================================

plt.figure(figsize=(8, 4))

plt.hist(clean["total_price"], bins=30)

plt.xlabel("Total price")
plt.ylabel("Number of orders")
plt.title("Distribution of total price")
plt.tight_layout()
plt.savefig("fig2.png", dpi=150)
plt.show()
```

## Plot 3 — `customer_rating` by category (clean data)

```python
# ============ TODO (Plot 3) ============
# One boxplot per category on a single axis.
# Hint: clean.boxplot(column='customer_rating', by='category')
# =======================================

plt.figure(figsize=(8, 4))

clean.boxplot(
    column="customer_rating",
    by="category"
)

plt.suptitle("")
plt.title("Customer rating by category")
plt.xlabel("Category")
plt.ylabel("Customer rating")
plt.tight_layout()
plt.savefig("fig3.png", dpi=150)
plt.show()
```
